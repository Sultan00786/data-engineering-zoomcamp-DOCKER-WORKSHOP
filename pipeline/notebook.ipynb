```python
# Read a sample of the data
prefix = 'https://github.com/DataTalksClub/nyc-tlc-data/releases/download/yellow/'
url = f'{prefix}/yellow_tripdata_2021-01.csv.gz'
url
```

```python
 # we need to specify the types:
dtype = {
    "VendorID": "Int64",
    "passenger_count": "Int64",
    "trip_distance": "float64",
    "RatecodeID": "Int64",
    "store_and_fwd_flag": "string",
    "PULocationID": "Int64",
    "DOLocationID": "Int64",
    "payment_type": "Int64",
    "fare_amount": "float64",
    "extra": "float64",
    "mta_tax": "float64",
    "tip_amount": "float64",
    "tolls_amount": "float64",
    "improvement_surcharge": "float64",
    "total_amount": "float64",
    "congestion_surcharge": "float64"
}

parse_dates = [
    "tpep_pickup_datetime",
    "tpep_dropoff_datetime"
]
```

```python
df = pd.read_csv(
    url,
    nrows=100,
    dtype=dtype,
    parse_dates=parse_dates
)
```

```python
df.head()
```

```python
df.info()
```

```python
# now need to insert data into PostgreSQL using SQLAlchemy
# Install SQLAlchemy
! uv add sqlalchemy
```

```python
from sqlalchemy import create_engine
engine = create_engine('postgresql+psycopg://root:root@localhost:5432/ny_taxi')
```

```python
engine
```

```python
# we want to create schema like this for 'yellow_taxi_data'
print(pd.io.sql.get_schema(df, name='yellow_taxi_data', con=engine))
```

```python
df.head(0)
```

```python
df.head(0).info()
```

```python
# Now create only schema as we see with zero values
df.head(0).to_sql(
    name='yellow_taxi_data', 
    con=engine,
    if_exists='replace' # if something is there then it gonna drop and replace with this empty value schema
)
```
